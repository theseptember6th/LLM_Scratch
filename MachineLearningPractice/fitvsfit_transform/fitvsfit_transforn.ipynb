In machine learning (especially with libraries like [scikit-learn](https://scikit-learn.org?utm_source=chatgpt.com)), `fit()` and `fit_transform()` are related but do different things.

### `fit()`

`fit()` **learns parameters from the data** but does **not return transformed data**.

Example:

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaler.fit(X_train)
```

What happens:

* Computes statistics from `X_train`

  * mean
  * standard deviation
* Stores them inside `scaler`
* Does **not change `X_train`**

You later use:

```python
X_scaled = scaler.transform(X_train)
```

---

### `fit_transform()`

`fit_transform()` does **two steps at once**:

1. `fit()` → learn from data
2. `transform()` → apply transformation

Example:

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_train)
```

Equivalent to:

```python
scaler.fit(X_train)
X_scaled = scaler.transform(X_train)
```

---

### Quick comparison

| Method            | Learns parameters? | Returns transformed data? |
| ----------------- | ------------------ | ------------------------- |
| `fit()`           | ✅ Yes              | ❌ No                      |
| `transform()`     | ❌ No               | ✅ Yes                     |
| `fit_transform()` | ✅ Yes              | ✅ Yes                     |

---

### Typical ML workflow

```python
scaler = StandardScaler()

# Training data
X_train_scaled = scaler.fit_transform(X_train)

# Test data
X_test_scaled = scaler.transform(X_test)
```

Why not `fit_transform(X_test)`?

Because that would compute new statistics from the test set → **data leakage**.

Training data teaches the transformation; test data only receives it.

`fit_predict()` is a convenience method in machine learning that **fits (trains) a model on data and immediately returns predictions in one step**.

Conceptually:

```text
fit_predict(X) = fit(X) + predict(X)
```

Where:

* `fit()` → learns patterns/parameters from the data
* `predict()` → generates outputs using what was learned

This method is especially common in **unsupervised learning** (clustering, anomaly detection).

### Example: Clustering with scikit-learn

```python
from sklearn.cluster import KMeans

X = [[1, 2], [1, 4], [10, 2], [10, 4]]

model = KMeans(n_clusters=2)

labels = model.fit_predict(X)

print(labels)
```

Output (example):

```text
[1 1 0 0]
```

What happened:

1. `fit(X)` → KMeans found cluster centers.
2. `predict(X)` → assigned each point to a cluster.
3. `fit_predict(X)` → did both together.

Equivalent code:

```python
model.fit(X)
labels = model.predict(X)
```

---

### Another example: Anomaly Detection

```python
from sklearn.ensemble import IsolationForest

model = IsolationForest()

result = model.fit_predict(X)
```

Possible output:

```text
[1, 1, -1, 1]
```

Meaning:

* `1` → normal point
* `-1` → anomaly

---

### When `fit_predict()` is typically available

* Clustering algorithms (KMeans, DBSCAN, Agglomerative Clustering)
* Some anomaly detection models
* Some dimensionality reduction or labeling workflows

### Difference from `predict()`

| Method           | What it does                |
| ---------------- | --------------------------- |
| `fit(X, y)`      | Train the model             |
| `predict(X)`     | Predict after training      |
| `fit_predict(X)` | Train + predict in one call |

For supervised learning (e.g., linear regression, random forest), the usual pattern is often:

```python
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

rather than `fit_predict()`.
