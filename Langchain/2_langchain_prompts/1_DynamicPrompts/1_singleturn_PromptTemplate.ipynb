### **2. What are Prompts?**

A **prompt** is the input message you send to a Language Model (LLM) to get a response. It can be text, an image, or audio (multi-modal), but this video focuses on **text-based prompts**.

*   **Why they are important:** An LLM's output is extremely sensitive to the prompt. A small change can lead to a vastly different result. This is why Prompt Engineering is a crucial skill.

---

### **3. Static vs. Dynamic Prompts**

#### **3.1 Static Prompts**
A static prompt is hard-coded by the developer. In a real application, you usually don't want this because you have less control over the user experience.

*   **Problem:** If you let the user write the entire prompt, they might write it poorly, misspell words, or ask for things in an inconsistent way. This leads to inconsistent responses from the LLM.

#### **3.2 Dynamic Prompts (and Prompt Templates)**
The solution is to create a **prompt template**. You ask the user for specific inputs (variables) and then fill them into a pre-written template. This ensures a consistent experience for all users.

*   **Example Scenario:** Building a research assistant that summarizes papers.
*   **User Inputs:** User selects `paper_name`, `explanation_style`, and `length`.
*   **Prompt Template:** A pre-written instruction with placeholders for these variables.

### **4. Prompt Templates (`PromptTemplate`)**

This class is used to create dynamic prompts for **single-turn** interactions.

*   **Why use `PromptTemplate` instead of Python f-strings?**
    1.  **Built-in Validation:** It automatically checks if all placeholders are filled, preventing runtime errors.
    2.  **Reusability:** You can save templates as JSON files and load them whenever needed.
    3.  **Integration:** It is tightly coupled with the LangChain ecosystem (e.g., for use in Chains).

```python
from langchain_core.prompts import PromptTemplate
```

```python
# 1.Define the template string
template=""" Please summarize the research paper titled "{paper_input}" with the following specifications:
    Explanation Style: {style_input}
    Explanation Length: {length_input}
    1. Mathematical Details:
        - Include relevant mathematical equations if present in the paper.
        - Explain the mathematical concepts using simple, intuitive code snippets where applicable.
    2. Analogies:
        - Use relatable analogies to simplify complex ideas.
    If certain information is not available in the paper, respond with: "Insufficient information available" instead of guessing.
    Ensure the summary is clear, accurate, and aligned with the provided style and length.
"""
```

```python
#2.Create the PromptTemplate Object
prompt_template=PromptTemplate(
    template=template,
    input_variables=["paper_input","style_input","length_input"],
    validate_template=True
)
```

```python
#3.Fill the template(invoke)
formatted_prompt=prompt_template.invoke(
    {
        "paper_input":"Attention is All You Need",
        "style_input":"code-oriented",
        "length_input":"medium"
    }
)

```

```python
# or if used invoke , u will get like /n and * and so on
print(formatted_prompt)
```

```python
type(formatted_prompt)
```

```python
formatted_prompt.text
```

```python
# if u  use the format instead of invoke,it cleans the output without any /n * and so on
formatted_prompt=prompt_template.format(
    paper_input="Attention is All You Need",
    style_input="code-oriented",
    length_input="medium"
)
```

```python
print(formatted_prompt)
```
