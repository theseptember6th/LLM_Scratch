### **6. Building a Chatbot and Message Types**

When building a multi-turn conversation (chatbot), you need to manage the conversation history. The video shows a console-based chatbot. The key challenge is that the LLM needs context from previous messages.

#### **6.1 The Problem: No Context**
If you only send the latest user message, the model forgets the conversation history.

#### **6.2 The Solution: Manage History with Message Types**
To give the model context, you must send the entire conversation history. But it's not enough to just send the text; you must label who said what. LangChain provides specific message classes for this:
1.  **`SystemMessage`**: A high-level instruction set at the beginning of a conversation (e.g., "You are a helpful assistant.").
2.  **`HumanMessage`**: A message from the user.
3.  **`AIMessage`**: A message from the AI.

## THE BELOW CODE WORKS

```python
from langchain_core.prompts import ChatPromptTemplate

# make tuples to make it work
# this also work (recommended)
chat_template = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a helpful {domain} expert"),
        ("human", "Explain in simple terms,what is {topic}"),
    ]
)
prompt = chat_template.invoke({"domain": "cricket", "topic": "Dusra"}) # this fills the variables dynamically

print(prompt)

```
