### **1. What are Chains and Why Do We Need Them?**

When building applications with Large Language Models (LLMs), your application is rarely a single step. It's usually made up of multiple smaller steps. For example, a simple app might involve:
1.  Taking a prompt from a user.
2.  Sending that prompt to an LLM.
3.  Processing the LLM's response to show it to the user.

Doing this manually for every step (creating a prompt, calling the model, parsing the output) can become tedious, especially for complex applications.

**Chains provide a solution.** A chain is a way to create a **pipeline** by connecting these individual steps together. The key feature is that the **output of one step automatically becomes the input for the next step**. You only need to provide the initial input to the first step, and the entire pipeline executes automatically.

**Benefits of using Chains:**
*   **Automation:** No need to manually call each component.
*   **Modularity:** Easy to build, connect, and reuse different components.
*   **Complexity:** Allows you to build complex, non-linear pipelines, not just simple sequences.

In this video, we explore three main types of chains:
*   **Sequential Chains:** Steps execute one after another in a series.
*   **Parallel Chains:** Multiple chains execute at the same time.
*   **Conditional Chains:** Different chains are executed based on a specific condition (like an if-else statement).

### **3. Building a Simple Sequential Chain**

This is the most basic type of chain. We'll build a three-step application that takes a topic from the user, asks an LLM to generate facts about it, and then parses the output to show only the text.

*   **Explanation:**
    1.  `PromptTemplate`: Creates the prompt with a variable (`topic`).
    2.  `Local AI`: The LLM model (e.g.,Mistral AI).
    3.  `StrOutputParser`: Extracts the string content from the model's response.
    4.  **Chain Creation:** We connect these components using the `|` operator. This is called the **LangChain Expression Language (LCEL)** . It's a declarative way to define your pipeline: `prompt | model | output_parser`.
    5.  **Invocation:** We call the chain with the initial input (`{"topic": "cricket"}`). The data flows automatically through each step.

### 1.Initialize the model and parser

```python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from dotenv import load_dotenv
import os

# Load environment variables
load_dotenv()

api_key = os.getenv("Hugging_face_api_token")

# Create LLM endpoint
llm = HuggingFaceEndpoint(
    # repo_id="TinyLlama/TinyLlama-1.1B-Chat-v1.0",
    repo_id="mistralai/Mistral-7B-Instruct-v0.2",
    task="text-generation",
    huggingfacehub_api_token=api_key,
)

# Wrap with chat interface
model = ChatHuggingFace(llm=llm)
```

```python
# Define the Output Parser
from langchain_core.output_parsers import StrOutputParser
parser=StrOutputParser()
```

### 2.Define the prompt template

```python
from langchain_core.prompts import PromptTemplate
template=PromptTemplate(
    template="Generate 5 interesting facts about {topic}",
    input_variables=["topic"]
)
```

### 3.Create the Chain by piping the components together and invoke

```python
chain= template | model | parser

result=chain.invoke({
    "topic":"Cricket"
})
```

### 4.Output

```python
print(result)
print(type(result))
```
