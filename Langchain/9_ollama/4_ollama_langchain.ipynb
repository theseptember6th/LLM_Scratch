## Quick Reference Commands

```bash
# Installation & Setup
ollama --version                    # Check installation
ollama pull model_name              # Download a model
ollama list                         # List downloaded models

# Running Models
ollama run model_name               # Start interactive session
ollama run model_name "prompt"      # One-shot prompt

# Management
ollama rm model_name                # Delete a model
ollama cp source destination        # Copy a model
ollama show model_name              # Show model details

# Cloud

ollama sign out                     # Sign out
```

## Topic 10: Using Ollama with LangChain

### Explanation
LangChain is an open-source framework for building advanced LLM applications. It acts as an **orchestrator layer** coordinating different components.

### Why LangChain?
Building production LLM applications requires handling multiple components:
- Input handling (PDFs, text files)
- Chunking
- Embeddings generation
- Vector databases
- Retrieval
- Generation

LangChain provides pre-built components for all these tasks.

### Three Ways to Use Ollama with LangChain

**1. Chat Interface** (Recommeneded - CHATMODELS)

# !pip install langchain-ollama

````python
from langchain_ollama import ChatOllama

# Initialize chat model
chat = ChatOllama(model='llama3.2:1b')

# Generate response
response = chat.invoke("Explain quantum entanglement in one sentence")
print(response.content)
````

**2. (NON_CHAT) only Text Generation (Completion) using LLM - Not recommended**

````python
from langchain_ollama import OllamaLLM

# Initialize LLM
llm = OllamaLLM(model='llama3.2:1b')

# Generate
response = llm.invoke("What is the capital of France?")
print(response)
````

**3. Embeddings Generation**

in CLI run this:


ollama pull nomic-embed-text

````python
from langchain_ollama import OllamaEmbeddings

# Initialize embeddings model
embeddings = OllamaEmbeddings(model='nomic-embed-text')

# Single text embedding
query_embedding = embeddings.embed_query("What is LangChain?")
print(query_embedding)

# Multiple documents embedding
doc_embeddings = embeddings.embed_documents([
    "Document 1 content",
    "Document 2 content"
])
print(doc_embeddings)
````
