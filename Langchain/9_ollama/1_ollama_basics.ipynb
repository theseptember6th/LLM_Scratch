## Topic 1: Introduction to LLMs (Large Language Models)

### Explanation
Large Language Models are complex neural networks with many layers and connections. They are essentially collections of numbers called **weights and biases**. All learning from training is stored in these weights and biases.

LLMs can be classified into two types based on accessibility:

**1. Proprietary Models**
- Owned and controlled by large companies (OpenAI, Google, etc.)
- You only get access to use the model via API or controlled platforms
- Architecture, weights, biases, training data are NOT publicly available
- Examples: GPT-4, Gemini, Claude
- Used through paid subscriptions

**2. Open Source Models**
- All components (architecture, weights, biases, training data) are made public
- Anyone can download and use them
- Can be run locally on your own system
- Can be fine-tuned and customized
- Examples: Llama (Meta), Mistral, DeepSeek, Qwen

### Key Point
While open source models are theoretically free, actually using them has significant friction points that Ollama solves.

## Topic 2: What is Ollama?

### Explanation
Ollama is a tool that helps you **download, run, and manage open source LLMs on your local computer**. It handles all the technical complexities:

- Storage optimization
- Working memory (RAM/VRAM) configuration
- Compatibility issues
- Request/response management

Think of Ollama as **"WhatsApp for open source models"** - you just focus on sending prompts and receiving responses; Ollama handles everything in the background.

### Benefits of Ollama
1. **Privacy & Data Control** - Data never leaves your computer
2. **Low Latency & Offline Access** - After initial download, you can use models without internet
3. **Cost Predictability** - No API costs; just electricity and hardware
4. **Simple Installation** - One-click setup
5. **Pre-built Model Library** - Hundreds of models readily available
6. **Customization** - Can modify parameters and create custom models

## Topic 5: Using Ollama - CLI Mode

### Explanation
Ollama can be used through command line interface (CLI) commands. This is primarily for experimentation and testing.

### Common CLI Commands

**1. Pull a model (download to local system)**
```bash
ollama pull llama3.2:1b # model name with parameters and version will be best
```

**2. List all downloaded models**
```bash
ollama ls
# or
ollama list
```

**3. Run a model (load into working memory)**
```bash
ollama run llama3.2:1b #model name with parameters and version will be best
```

**4. Delete a model**
```bash
ollama rm model_name
```

### Running a Model - Interactive Mode
```bash
ollama run llama3.2:1b
```
Once loaded, you can:
- Ask questions directly
- Use `/bye` to exit
- Use `/show` to view model information
- Use `/set` to modify parameters

### Model Information Commands (inside running model)
```
/show info      - Show model architecture, parameters, context length
/show parameters - Show model parameters
/show system    - Show system instructions
/set parameter top_p 0.90  - Change a parameter
/set system "You are a helpful assistant" - Set system instruction
```

### Vision Models
Some models have vision capabilities. To use vision models:
```bash
ollama run gemma3:4b
```
Then provide image path with your prompt:
```
Summarize the image [image_path]
```

- if stream = True print response continously as it goes 
- if stream =False ,print response only at last after it finishes.

````python
stream = ollama.generate(
    model='mistral:7b',# model  i downloaded locally using cli command,ollama pull mistral:7b
    prompt='Why does the moon glow?',
    stream=True # responses is shown continuously as it generates
)

for chunk in stream:
    print(chunk['response'], end='')
````

### Passing Images (Base64 Encoded) as per documentation (Check endpoints)

<img src="api_reference.png">

````python
## Giving Multimodel input to model ( pasing image to the model)
import ollama
import base64

# Encode image to base64
with open("./cli.png","rb") as f: # Opening the cli Image in Binary
    image_bytes=f.read() # reading the image
    image_64=base64.b64encode(image_bytes).decode("utf-8") #encoding the image in base 64 bits


response=ollama.generate(
    model="gemma3:4b", # model for image i downloaded locally using cli command,ollama pull gemma3:4b
    images=[image_64], #must be a list of base 64 bits encoded image
    prompt="Give caption for the image"

)

print(response["response"])
````

### Multiple Images

````python
# PASSING MULTIPLE IMAGES TO THE MODEL.

image_paths=["./cli.png","./api_reference.png"]

# Encode multiple images
images_base64=[]

for image_path in image_paths:
    with open(image_path,"rb") as f:
        image_bytes=f.read() # reading the image
        image_64=base64.b64encode(image_bytes).decode("utf-8") #encoding the image in base 64 bits
        images_base64.append(image_64)


response=ollama.generate(
    model="gemma3:4b", # model for image i downloaded locally using cli command,ollama pull gemma3:4b
    images=images_base64, #must be a list of base 64 bits encoded image
    prompt="Give caption for the image"

)

print(response.response)
````

### System Instructions to the  models using ollama

````python
response=ollama.generate(
    model="mistral:7b", # ur locally downloaded model by using cli, ollama pull mistral:7b
    prompt="Why does the moon glow?",
    system="You are a funny assistant,you explain things in funny way" #system instruction
)

print(response.response)
````

### Custom Parameters (change,modify parameters) of model

````python
response=ollama.generate(
    model="mistral:7b", # ur locally downloaded model by using cli, ollama pull mistral:7b
    prompt="Why is  the ocean blue?",
    options={    #change,modify parameters
        "temperature":0.3,
        "top_p":0.5,
        "top_k":45
    }
)

print(response.response)
````

## Topic 7: Using Ollama - REST API

### Explanation
Ollama internally works using REST API. All CLI commands and Python library are just wrappers around API endpoints.

### Local Server
When you run Ollama, it creates a local server at: `http://localhost:11434`

### API Endpoints

**1. Generate (POST)**

````python
import requests
import json

url = "http://localhost:11434/api/generate"

payload={
    "model":"mistral:7b",
    "prompt":"Why is sky blue?"
}

response=requests.post(url,json=payload,stream=True)

for line in response.iter_lines():
    if line:
        data = json.loads(line)
        print(data['response'], end='')
````

**2. List Models (GET)**

````python
url = "http://localhost:11434/api/tags"
response = requests.get(url)
models = response.json()
print(models)
````

## Generate a chat message

- Generate the next chat message in a conversation between a user and an assistant.



Note: Generate was only used for single turn message not for conversation

````python

import requests
import json

url="http://localhost:11434/api/chat"

payload={
    "model":"mistral:7b",
    "messages": [
    {
      "role": "user",
      "content": "why is the sky blue?"
    }
  ]
}

response=requests.post(url,json=payload,stream=True)
response
````
