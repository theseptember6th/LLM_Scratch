## Quick Reference Commands

```bash
# Installation & Setup
ollama --version                    # Check installation
ollama pull model_name              # Download a model
ollama list                         # List downloaded models

# Running Models
ollama run model_name               # Start interactive session
ollama run model_name "prompt"      # One-shot prompt

# Management
ollama rm model_name                # Delete a model
ollama cp source destination        # Copy a model
ollama show model_name              # Show model details

# Cloud
ollama sign in                      # Sign in to Ollama Cloud
ollama sign out                     # Sign out
```

## Topic 9: Model Files (Custom Models)

### Explanation
Model files are configuration files that define how an existing base model should behave. They contain **instructions, not the model itself**.

### What Model Files Do
- Take an existing trained model (base model)
- Add instructions to customize its behavior
- Create a new model that uses the base model's learning in a specific way

### Model File Structure
```
FROM llama3.2:1b          # Base model to use

PARAMETER temperature 0.7
PARAMETER top_p 0.9

SYSTEM """You are a sentiment analysis API. 
You only output JSON in the following schema:
{
    "sentiment": "positive|negative|neutral",
    "score": 0.0 to 1.0
}
"""

# Optional: Provide examples
MESSAGE user "I love this product!"
MESSAGE assistant {"sentiment": "positive", "score": 0.95}
```

### Creating a Custom Model (CLI WAY)

1. **Save your model file** (e.g., `model_file.txt`)

2. **Create the custom model**
```bash
ollama create sentiment:latest -f ./model_file.txt
```

3. **Use the custom model**
```bash
ollama run sentiment:latest
```
Then any prompt will follow your custom instructions.

### Example: Sentiment Analysis Model
```bash
# Create model name "sentiment:latest" or any name u want
ollama create sentiment:latest -f ./sentiment_model.txt

# Use it
ollama run sentiment:latest
>>> I love this course
{"sentiment": "positive", "score": 0.95}
```
