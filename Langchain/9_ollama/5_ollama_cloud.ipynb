## Quick Reference Commands

```bash
# Installation & Setup
ollama --version                    # Check installation
ollama pull model_name              # Download a model
ollama list                         # List downloaded models

# Running Models
ollama run model_name               # Start interactive session
ollama run model_name "prompt"      # One-shot prompt

# Management
ollama rm model_name                # Delete a model
ollama cp source destination        # Copy a model
ollama show model_name              # Show model details

# Cloud
ollama sign in                      # Sign in to Ollama Cloud
ollama sign out                     # Sign out
```

## Topic 11: Ollama Cloud

### Explanation
Ollama Cloud is a cloud-based extension that allows running large parameter models without needing powerful local hardware.

### The Problem
- Large models (100B+ parameters) require massive computational resources
- Most local hardware can't run these models
- Even though models are downloadable, hardware constraints prevent usage

### The Solution
Ollama Cloud provides:
- Data center-grade hardware
- Run large models (like DeepSeek V3 with 671B parameters)
- No local hardware constraints

### How to Use Ollama Cloud

1. **Sign in to Ollama Cloud**
```bash
ollama sign in
# Follow the URL provided to authenticate
```

2. **Run a cloud model**
```bash
# Models with cloud capability have the cloud tag
ollama run deepseek-v3:671b
```

3. **Verify connection**
```
Connected to deepseek-v3:671b (cloud)
>>> What are the colors of the rainbow?
```

### Important Considerations

| Aspect | Local Models | Cloud Models |
|--------|--------------|--------------|
| Hardware | Your local system | Data center |
| Model Size | Limited by hardware | Any size available |
| Privacy | Complete | Data leaves your system |
| Internet | Not required after download | Always required |
| Cost | Free | Free tier (limited requests) |

### Privacy Note
Ollama claims they don't retain request data - they process and delete immediately. However, data must travel from your system to their servers, so there's always some privacy trade-off.
