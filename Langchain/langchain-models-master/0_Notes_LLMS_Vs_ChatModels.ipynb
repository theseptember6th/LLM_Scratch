

## **LangChain Models: LLMs vs. Chat Models - Detailed Notes**

### **Topic 1: The Core Distinction**

**Explanation:**
According to the video, Language Models in LangChain are divided into two types: **LLMs (Legacy)** and **Chat Models (Recommended)** . The main difference lies in their design purpose and how they handle input and output.

*   **LLMs (Large Language Models):** These are **general-purpose models**. They are designed for a wide range of NLP tasks like text generation, summarization, translation, and code generation. They take a **plain string** as input and return a **plain string** as output. The instructor notes that LangChain is gradually phasing out support for this interface, recommending against it for new projects.
*   **Chat Models:** These are **specialized for conversation**. They are fine-tuned on conversational data. They take a **sequence of messages** (with roles like system, user, assistant) as input and return a **chat message** as output. They natively understand context and conversation history.

---

### **Topic 2: Detailed Comparison Table (as per Video)**

The instructor provides a clear comparison, which is summarized below:

| Feature | LLM | Chat Model |
| :--- | :--- | :--- |
| **Purpose** | Free-form text generation (any NLP task). | Multi-turn conversations, chat bots, agents. |
| **Training** | Trained on general text (books, articles, Wikipedia). | Trained on general text + fine-tuned on chat datasets. |
| **Input Type** | A single string. | A list of messages (with roles). |
| **Output Type** | A single string. | A chat message (with metadata). |
| **Memory** | No concept of conversation history. | Supports conversation history. |
| **Role Awareness** | Cannot assign roles (like "system" or "user"). | Understands roles (system, user, AI). |
| **Example Models** | `gpt-3.5-turbo-instruct` | `gpt-4`, `claude-3`, `gemini-1.5-pro`, `TinyLlama-Chat` |
| **When to Use** | Older, general text tasks. | **Modern AI applications**: chat bots, virtual assistants, agents. |

### **Topic 3: LLM Code Demo (OpenAI)**

**Explanation:**
This demonstrates the legacy LLM interface. Notice how the input is a simple string, and the output is also a simple string. The instructor uses the `OpenAI` class (not `ChatOpenAI`) and the `gpt-3.5-turbo-instruct` model.

**Code:**
```python
import os
from dotenv import load_dotenv
from langchain_openai import OpenAI

# Load API key from .env file
load_dotenv()

# Create an LLM instance (Legacy)
llm = OpenAI(model="gpt-3.5-turbo-instruct")

# Invoke with a simple string prompt
result = llm.invoke("What is the capital of India?")

# Output is a plain string
print(result)
```
**Output:**
```
The capital of India is New Delhi.
```
**Instructor's Note:**
> *"यह जो कोड आप देख रहे हो यह शायद आपको कुछ पुराने मॉडल्स... यह अभी सबसे इंपॉर्टेंट और सबसे रिलेवेंट पीस ऑफ कोड होंगे... यहां पर भी आप नोटिस करो कि इस एलएलएम को हमने एक स्ट्रिंग भेजा है और उसने पलट के हमें जो भेजा है वह भी एक स्ट्रिंग है।"*
> *(Translation: You might find this code for older models... but this is important and relevant... Notice here that we sent a string to the LLM and it sent back a string.)*

### **Topic 4: Chat Model Code Demo (OpenAI - GPT-4)**

**Explanation:**
This is the modern, recommended approach. The key difference is the import: `ChatOpenAI` instead of `OpenAI`. The output is no longer a simple string but an object containing the content and metadata. You must access `.content` to get the actual text.

**Code:**
```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI

load_dotenv()

# Create a Chat Model instance
model = ChatOpenAI(model="gpt-4")

# Invoke with a prompt
result = model.invoke("What is the capital of India?")

# The result is an object; print the whole object first to see the metadata
print("Full Result Object:", result)

# To get just the answer, access the 'content' attribute
print("\nAnswer:", result.content)
```
**Output (Partial - showing object structure):**
```
Full Result Object: content='The capital of India is New Delhi.' response_metadata={'token_usage': {'completion_tokens': 7, 'prompt_tokens': 14, 'total_tokens': 21}, 'model_name': 'gpt-4', ...}

Answer: The capital of India is New Delhi.
```
**Instructor's Note:**
> *"अब एक चीज आप नोटिस करोगे कि पिछले बार की तरह यहां पर जो रिजल्ट है वह सिंपल प्लेन टेक्स्ट नहीं है... सबसे पहले आपको कंटेंट बोल कर के की दिखाई दे रहा है जिसके अंदर आपका एक्चुअल आंसर छुपा हुआ है।"*
> *(Translation: Now you will notice that unlike last time, the result here is not simple plain text... First, you see something called 'content' which contains your actual answer.)*

### **Topic 5: Key Parameter - Temperature (for Chat Models)**

**Explanation:**
The `temperature` parameter controls the randomness/creativity of the model's output. The video provides a practical guide for selecting values:

*   **0.0 – 0.3:** **Deterministic.** Best for factual answers, coding, math.
*   **0.5 – 0.7:** **Balanced.** Good for general Q&A and explanations.
*   **0.9 – 1.2:** **Creative.** Good for writing stories, poems, jokes.
*   **1.5+:** **Very Random.** Useful for brainstorming.

**Code (Demonstrating Temperature):**
```python
from langchain_openai import ChatOpenAI

# Low temperature - predictable
model_low_temp = ChatOpenAI(model="gpt-4", temperature=0.1)
result_low = model_low_temp.invoke("Suggest me five Indian male names.")
print("Low Temp (0.1):", result_low.content)

# High temperature - more varied
model_high_temp = ChatOpenAI(model="gpt-4", temperature=1.5)
result_high = model_high_temp.invoke("Suggest me five Indian male names.")
print("High Temp (1.5):", result_high.content)
```

### **Topic 6: Key Parameter - Max Tokens**

**Explanation:**
`max_tokens` (or `max_completion_tokens`) limits the length of the model's response. This is important for controlling costs, as you pay per token. The video explains that tokens are roughly equivalent to words for understanding purposes.

**Code:**
```python
from langchain_openai import ChatOpenAI

# Limit the output to 10 tokens
model = ChatOpenAI(model="gpt-4", max_tokens=10)

result = model.invoke("Tell me about cricket.")
print(result.content)
```
**Output (will be a very short, cut-off sentence):**
```
Cricket is a bat-and-ball game played between
```
