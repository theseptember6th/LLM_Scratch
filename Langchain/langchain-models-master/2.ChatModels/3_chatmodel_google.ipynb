### **Topic 4: Chat Model Code Demo (ChatGoogleGenerativeAI)**

**Explanation:**
This is the modern, recommended approach. The key difference is the import: `ChatGoogleGenerativeAI`->Chat Model instead of `GoogleGenerativeAI`->LLM. The output is no longer a simple string but an object containing the content and metadata. You must access `.content` to get the actual text.

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
import os


load_dotenv()
api_key = os.getenv("GOOGLE_API_KEY")

# Create a Chat Model instance
model = ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite", api_key=api_key)

# Invoke with a prompt
result = model.invoke("What is the capital of Kathmandu?")

# The result is an object; print the whole object first to see the metadata
print(result)
```

```python
# To get just the answer, access the 'content' attribute
result.content
```
